## 3. Data Cleaning and Type Conversion

```python
df.isnull().sum()
```

```python
count_columns=[
    'Searches',
    "Searches which got estimate",
    "Searches for Quotes",
    "Searches which got Quotes",
    "Bookings",
    "Completed Trips",
    "Cancelled Bookings",
    "Distance Travelled (km)"
]

for col in count_columns:
  df[col] = df[col].astype(str).str.replace(",","", regex = False).astype(int)
```

```python
precent_columns=[
    "Search-to-estimate Rate",
    "Estimate-to-search for quotes Rate",
    "Quote Acceptance Rate",
    "Quote-to-booking Rate",
    "Booking Cancellation Rate",
    "Conversion Rate"
]

for col in precent_columns:
  df[col] = df[col].astype(str).str.replace("%","", regex = False).astype(float)
```

```python
df["Drivers' Earnings"] = (
    df["Drivers' Earnings"]
      .astype(str)
      .str.replace ("₹","", regex=False)
      .str.replace (",", "", regex=False)
      .astype(float)
 )
```

```python
df["Average Fare per Trip"].head()
```

```python
df["Average Fare per Trip"] = (
    df["Average Fare per Trip"]
      .astype(str)
      .str.replace("₹", "", regex=False)
      .str.replace(",", "", regex=False)
      .astype(float)
)
```

```python
df["Average Fare per Trip"] = df["Average Fare per Trip"].astype(float)
```

## 8. Key Business Insights

Based on the comprehensive analysis of the driver partner data for Bangalore Wards, here are the key business insights:

*   **Strong Correlation in Core Metrics:** 'Searches', 'Bookings', 'Completed Trips', and 'Drivers' Earnings' are highly correlated, indicating that increased activity at the top of the funnel directly drives driver income.
*   **Data Quality is High:** The dataset is complete with no missing values, and data types were successfully converted, providing a robust foundation for analysis.
*   **Variability in Conversion Rates:** 'Conversion Rate' ranges significantly (14.6% to 46.4%) across wards, pointing to opportunities for targeted improvements in lower-performing areas.
*   **High-Performing Wards Identified:** Wards like 'Other Wards', 'Shantala Nagar', 'Doddakanahalli', 'Agara', and 'Koramangala' consistently lead in 'Searches', 'Completed Trips', and 'Drivers' Earnings', marking them as crucial for business volume.
*   **Cancellation Rates Need Attention:** 'Booking Cancellation Rate' varies widely (24.1% to 62.3%). Wards with high cancellation rates should be investigated to address underlying issues.
*   **Outliers in Performance:** Boxplots revealed outliers in 'Drivers' Earnings', 'Searches', and 'Completed Trips', suggesting unique operational scenarios or high-impact drivers/wards that warrant further examination.
*   **Strategic Implications for Growth:** Interventions could focus on boosting conversion in weaker wards, optimizing operations in high-cancellation areas, and leveraging insights from high-performing wards for overall growth.
